```python
import pandas as pd
import matplotlib.pyplot as plt

df = {
    "Roll": [1, 2, 3, 4, 5],
    "Python": [45, 23, 33, 50, 43],
    "AI": [55, 69, 14, 39, 78]
}

df = pd.DataFrame(df)

df.plot(x="Roll", y=["Python", "AI"], kind="bar")

plt.title("Python and AI Marks")
plt.xlabel("Roll Number")
plt.ylabel("Marks")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

BCA = [128, 185, 295, 348, 305, 398]
MCA = [20, 35, 45, 65, 95, 110]
Years = [2021, 2022, 2023, 2024, 2025, 2026]

plt.plot(Years, BCA, marker='s', label='BCA-Admission')
plt.plot(Years, MCA, marker='^', label='MCA-Admission')

plt.grid(True)
plt.title('Admission Analysis')
plt.xlabel('Years')
plt.ylabel('Number of Admissions')

plt.legend()

plt.savefig("Admission_data.png")
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.DataFrame({
    "Name": ["A", "B", "C", "D", "E"],
    "Roll": [1, 2, 3, 4, 5],
    "Age": [20, 35, 22, 41, 31],
    "Gender": ["Male", "Female", "Male", "Female", "Male"]
})

ax=sns.barplot(data=df, x="Name", y="Age", hue="Gender") #in  bar we are not share marker
plt.title('Age Analysis')
plt.xlabel('Name')
plt.ylabel('Age')
plt.grid()
for i in ax.containers:
    ax.bar_label(i)
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.DataFrame({
    "Name": ["A", "B", "C", "D", "E"],
    "Roll": [1, 2, 3, 4, 5],
    "Age": [20, 35, 22, 41, 31],
    "Gender": ["Male", "Female", "Male", "Female", "Male"]
})

ax = sns.lineplot(
    data=df, x="Name", y="Age",
    marker="o", hue="Gender"
)

plt.title("Age Analysis - Line Graph")
plt.xlabel("Name")
plt.ylabel("Age")
plt.grid(True)

for i in range(len(df)):
    plt.text(i, df["Age"][i] + 1, str(df["Age"][i]),
             ha="center")

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.DataFrame({
    "Name": ["A", "B", "C", "D", "E"],
    "Roll": [1, 2, 3, 4, 5],
    "Age": [20, 35, 22, 41, 31],
    "Gender": ["Male", "Female", "Male", "Female", "Male"]
})

ax = sns.scatterplot(
    data=df, x="Roll", y="Age",
    hue="Gender", s=100
)

plt.title("Age Analysis - Scatter Plot")
plt.xlabel("Roll Number")
plt.ylabel("Age")
plt.grid(True)

for i in range(len(df)):
    plt.text(df["Roll"][i], df["Age"][i] + 1,
             str(df["Age"][i]), ha="center")

plt.show()
```
